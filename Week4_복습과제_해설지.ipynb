## 데이터 준비 — 그대로 실행

위스콘신 유방암 데이터의 레이블은 `0=악성`, `1=양성`입니다. 전체 데이터에서 학습용 80%와 테스트용 20%를 분리하고, 학습용을 다시 학습·검증 데이터로 나눕니다.


```python
import numpy as np
import xgboost as xgb
from xgboost import XGBClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

cancer = load_breast_cancer()
X, y = cancer.data, cancer.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=156, stratify=y
)
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train, test_size=0.1, random_state=156, stratify=y_train
)

print('전체 데이터:', X.shape)
print('학습:', X_tr.shape, '검증:', X_val.shape, '테스트:', X_test.shape)
print('레이블:', dict(enumerate(cancer.target_names)))
```

## 문제 1. GBM 모델 학습과 정확도 평가

`GradientBoostingClassifier`로 유방암 데이터를 분류한다.

1. `n_estimators=100`, `learning_rate=0.1`, `max_depth=3`, `random_state=42`로 모델 생성
2. **`X_train`, `y_train`**으로 학습하고 `X_test` 예측
3. 테스트 정확도를 소수점 넷째 자리까지 출력


```python
# 약한 학습기 100개를 순차적으로 학습하는 GBM 분류기
gbm = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# 검증용으로 나누기 전의 학습 데이터 455건을 사용
gbm.fit(X_train, y_train)
gbm_pred = gbm.predict(X_test)

gbm_acc = accuracy_score(y_test, gbm_pred)
print('GBM 테스트 정확도: {:.4f}'.format(gbm_acc))
```

### 문제 1 풀이 설명

`GradientBoostingClassifier`는 사이킷런의 GBM 분류기이다. `n_estimators=100`은 순차적으로 학습할 약한 학습기의 개수이다. `learning_rate=0.1`은 각 약한 학습기의 보정을 모델에 반영하는 정도다. `max_depth=3`은 개별 트리의 깊이를 제한하고, `random_state=42`는 난수를 사용하는 과정을 재현할 수 있게 한다.

`fit(X_train, y_train)`으로 학습용 데이터 455건을 사용한다. `predict(X_test)`는 테스트 데이터 114건의 클래스를 예측한다. `accuracy_score(y_test, gbm_pred)`는 정답과 예측 클래스가 일치한 비율을 계산한다. `{:.4f}`는 계산한 정확도를 소수점 넷째 자리까지 표시한다.


## 문제 2. 파이썬 래퍼용 `DMatrix` 만들기

학습·검증·테스트 데이터를 `xgb.DMatrix`로 각각 변환한다.

1. **`X_tr`, `y_tr`**로 `dtr` 생성
2. **`X_val`, `y_val`**로 `dval` 생성
3. **`X_test`, `y_test`**로 `dtest` 생성
4. `num_row()`를 이용해 각각의 데이터 건수 출력


```python
# 파이썬 래퍼에서 사용할 XGBoost 전용 데이터 객체
dtr = xgb.DMatrix(data=X_tr, label=y_tr)
dval = xgb.DMatrix(data=X_val, label=y_val)
dtest = xgb.DMatrix(data=X_test, label=y_test)

print('학습:', dtr.num_row())
print('검증:', dval.num_row())
print('테스트:', dtest.num_row())
```

### 문제 2 풀이 설명

파이썬 래퍼의 `xgb.train()`은 XGBoost 전용 데이터 객체인 `DMatrix`를 입력받는다. `data`에는 피처를, `label`에는 그 피처에 대응하는 정답을 넣는다. 따라서 `X_tr`에는 `y_tr`, `X_val`에는 `y_val`, `X_test`에는 `y_test`를 연결한다.

`dtr`은 모델 학습에, `dval`은 학습 중 검증에, `dtest`는 최종 테스트 예측에 사용한다. `num_row()`는 각 객체에 담긴 데이터 건수를 반환한다. 준비 셀의 분할에 따라 **학습 409건, 검증 46건, 테스트 114건**이 출력된다. 테스트 레이블은 이 문제에서 `DMatrix`에 넣었지만, 문제 4의 예측 확률 계산 자체에는 사용하지 않는다.


## 문제 3. 파이썬 래퍼 XGBoost 학습과 조기 중단

문제 2의 `dtr`과 `dval`로 모델을 학습한다.

1. `params` 딕셔너리에 `max_depth=3`, `eta=0.05`, `objective='binary:logistic'`, `eval_metric='logloss'`, `seed=42` 지정
2. 평가 목록 `[(dtr, 'train'), (dval, 'eval')]`을 만들어 학습·검증 데이터 지정
3. `xgb.train()`에 `num_boost_round=200`, `early_stopping_rounds=20` 적용
4. 학습 완료 후 모델의 `best_iteration`과 `best_score` 출력

학습 로그가 너무 길면 `verbose_eval=False`를 추가할 수 있다.


```python
# 파이썬 래퍼의 학습 설정: 트리 깊이, 학습률, 목적 함수, 검증 지표
params = {
    'max_depth': 3,
    'eta': 0.05,
    'objective': 'binary:logistic',
    'eval_metric': 'logloss',
    'seed': 42
}

# 검증용 dval을 마지막에 배치해 조기 중단의 기준으로 사용
eval_list = [(dtr, 'train'), (dval, 'eval')]

xgb_model = xgb.train(
    params=params,
    dtrain=dtr,
    num_boost_round=200,
    evals=eval_list,
    early_stopping_rounds=20,
    verbose_eval=False
)

print('가장 좋은 검증 점수를 낸 반복 번호:', xgb_model.best_iteration)
print('해당 반복의 검증 점수:', xgb_model.best_score)
```

### 문제 3 풀이 설명

파이썬 래퍼에서는 학습 설정을 `params` 딕셔너리로 전달한다. `max_depth=3`은 트리의 최대 깊이를 정하고, `eta=0.05`는 각 부스팅 단계의 학습률을 정한다. `objective='binary:logistic'`은 이진 분류의 목적 함수를 지정한다. `eval_metric='logloss'`는 검증에 사용할 손실 지표이며, 값이 낮을수록 좋다. `seed=42`는 XGBoost에서 사용하는 난수 시드다.

`eval_list`의 각 항목은 `(DMatrix, 데이터 이름)` 형식이다. 학습용 `dtr`을 `train`, 검증용 `dval`을 `eval`로 표시한다. 검증용 데이터를 마지막에 두었으므로 `early_stopping_rounds=20`은 검증 `logloss`의 개선 여부를 기준으로 작동한다. `num_boost_round=200`은 최대 반복 횟수이며, 검증 점수가 20회 동안 개선되지 않으면 그전에 학습이 종료된다.

`best_iteration`은 검증 `logloss`가 가장 낮았던 반복 번호다. 반복 번호는 0부터 시작한다. `best_score`는 해당 반복의 검증 `logloss`다. 두 값은 학습 결과를 확인하기 위해 출력한다.


## 문제 4. 예측 확률을 클래스로 바꾸고 평가하기

문제 3의 파이썬 래퍼 모델로 테스트 데이터를 예측한다.

1. `predict(dtest)`로 **클래스 1의 예측 확률** 계산
2. 확률이 **0.5보다 크면 1**, 그렇지 않으면 **0**으로 변환
3. 예측 확률 앞 5개와 변환한 클래스 앞 5개 출력
4. `accuracy_score()`로 테스트 정확도를 소수점 넷째 자리까지 출력

확률을 클래스로 변환할 때는 리스트 컴프리헨션이나 `np.where()` 중 편한 방법을 사용한다.


```python
# binary:logistic 모델의 예측값: 클래스 1의 예측 확률
pred_probs = xgb_model.predict(dtest)

# 확률이 0.5보다 큰 경우에만 클래스 1로 변환
native_preds = np.where(pred_probs > 0.5, 1, 0)
native_acc = accuracy_score(y_test, native_preds)

print('예측 확률 앞 5개:', np.round(pred_probs[:5], 3))
print('변환한 클래스 앞 5개:', native_preds[:5])
print('파이썬 래퍼 테스트 정확도: {:.4f}'.format(native_acc))
```

### 문제 4 풀이 설명

`binary:logistic`으로 학습한 모델의 `predict(dtest)`는 클래스 1에 속할 예측 확률을 반환한다. 이 데이터에서 클래스 1은 양성 종양이다. 확률이 0.5보다 크면 클래스 1, 그렇지 않으면 클래스 0으로 정한다. `np.where(pred_probs > 0.5, 1, 0)`이 이 변환을 수행한다.

`pred_probs`에는 확률이, `native_preds`에는 0 또는 1의 예측 클래스가 저장된다. `np.round(pred_probs[:5], 3)`은 앞의 확률 5개를 소수점 셋째 자리까지 표시한다. 정확도는 예측 클래스 `native_preds`와 실제 클래스 `y_test`를 `accuracy_score()`로 비교해 계산한다.

# 문제 9. 성능 평가하기
교재에서 사용된 `get_clf_eval` 함수입니다. 함수를 실행하고, 그 아래 코드에 알맞은 파라미터를 넣어 성능 평가를 수행해주세요.

```python
from sklearn.metrics import (accuracy_score, precision_score,
  recall_score, f1_score, roc_auc_score, confusion_matrix)

def get_clf_eval(y_test, pred=None, pred_proba=None):
  confusion = confusion_matrix(y_test, pred)
  accuracy = accuracy_score(y_test, pred)
  precision = precision_score(y_test, pred)
  recall = recall_score(y_test, pred)
  f1 = f1_score(y_test, pred)
  roc_auc = roc_auc_score(y_test, pred_proba)
  print('오차 행렬')
  print(confusion)
  print('정확도:{:.4f}, 정밀도: {:.4f}, 재현율:{:.4f},\
  F1:{:.4f}, AUC: {:.4f}'.format(accuracy, precision, recall, f1, roc_auc))

get_clf_eval(y_test, preds, pred_proba)
```

**LightGBM은 일반 GBM 계열의 트리 분할 방법과 다르게 어떤 방식을 사용하나요? 그 방식의 장단점을 간략히 서술하세요.**

LightGBM은 일반 GBM 계열 알고리즘이 사용하는 균형 트리 분할(Level Wise) 방식 대신, 리프 중심 트리 분할(Leaf Wise) 방식을 사용한다.
균형 트리 분할은 트리의 깊이를 최소화하기 위해 최대한 균형 잡힌 트리를 유지하면서 분할하는 방식으로, 오버피팅에 좀 더 강한 구조를 가질 수 있지만 균형을 맞추기 위한 시간이 필요하다는 단점이 있다.
반면 리프 중심 트리 분할은 트리의 균형을 맞추지 않고, 최대 손실 값(max delta loss)을 가지는 리프 노드를 지속적으로 분할하면서 트리의 깊이가 깊어지고 비대칭적인 규칙 트리가 생성된다.
이렇게 최대 손실값을 가지는 리프 노드를 지속적으로 분할해 생성된 규칙 트리는 학습을 반복할수록 균형 트리 분할 방식보다 예측 오류 손실을 최소화할 수 있다는 것이 LightGBM의 구현 사상이다.
다만 적은 데이터 세트(일반적으로 10,000건 이하)에 적용할 경우 과적합이 발생하기 쉽다는 단점이 있다.

**베이지안 최적화가 Grid Search 방식에 비해 갖는 장점을 대체 모델과 획득 함수의 역할을 포함하여 서술하세요.**

Grid Search 방식은 지정한 모든 하이퍼 파라미터 값의 조합을 순차적으로 하나씩 학습하고 평가하는 방식이기 때문에 튜닝해야 할 하이퍼 파라미터의 개수가 많거나, 값의 범위가 넓을 경우 조합의 개수가 기하급수적으로 늘어나 수행 시간이 매우 오래 걸린다는 단점이 있다.
베이지안 최적화는 목적 함수의 식 자체를 알지 못하는 블랙박스 상황에서도, 적은 시도 횟수만으로 최적 입력값을 효율적으로 찾아낼 수 있다.
이를 위해 대체 모델(Surrogate Model)과 획득 함수(Acquisition Function)라는 두 가지 요소를 사용한다.
대체 모델은 지금까지 관측된 (하이퍼 파라미터, 성능) 값들을 바탕으로 목적 함수의 형태를 추정하며, 새로운 샘플 데이터가 들어올 때마다 추정을 갱신하며 점차 정확해진다.
획득 함수는 대체 모델이 추정한 함수를 바탕으로, 이전보다 더 좋은 성능을 낼 가능성이 높은 다음 하이퍼 파라미터 지점을 계산해 대체 모델에 제안한다.
이 두 요소가 서로 상호작용하며 반복적으로 개선되기 때문에, Grid Search처럼 모든 조합을 다 시도하지 않고도 적은 횟수의 시도로 최적값에 가까운 하이퍼 파라미터를 빠르게 찾을 수 있다.

## 문제 17. IQR 방식을 구현한 이상치 제거

신용카드 사기 데이터에서 사기(Fraud, Class=1)와 정량적 상관관계가 높은 특정 피처(예: V14)의 이상치를 IQR 방식을 통해 제거하는 함수 get_outlier()의 일부이다. 사분위수(Percentile)를 이용해 IQR을 구하고 최솟값과 최댓값을 설정하는 빈칸에 들어갈 올바른 코드는?


```python
import numpy as np
def get_outlier(df=None, column=None, weight=1.5):
    fraud = df[df['Class']==1][column]
    quantile_25 = np.percentile(fraud.values, 25)
    quantile_75 = np.percentile(fraud.values, 75)
    
    iqr = quantile_75 - quantile_25
    iqr_weight = iqr * weight
    lowest_val = quantile_25 - iqr_weight
    hightest_val = quantile_75 + iqr_weight
    
    outlier_index = fraud[(fraud < lowest_val) | (fraud > hightest_val)].index
    return outlier_index
```

해설 : IQR(사분위수 범위)에 가중치(1.5)를 곱한 값을 Q1에서 빼고 Q3에 더하여 각각 최솟값과 최댓값을 구한 뒤, 이를 벗어나는 데이터 인덱스를 이상치로 판별합니다.
