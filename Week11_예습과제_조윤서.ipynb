```python
def visualize_cluster_2d(df, cluster_col, centers=None, legend=True):
    from sklearn.decomposition import PCA

    X = df[['sepal_length', 'sepal_width',
            'petal_length', 'petal_width']].values

    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)

    markers = ['o', 's', '^', 'x', '*', 'D', 'P']
    unique_labels = np.unique(df[cluster_col])

    for label in unique_labels:
        mask = df[cluster_col] == label

        plt.scatter(
            x=X_pca[mask, 0],
            y=X_pca[mask, 1],
            marker='+' if label == -1 else markers[int(label) % len(markers)],
            edgecolor='k',
            color='k' if label == -1 else None,
            label=f'Cluster {label}'
        )

    if centers is not None:
        for i, center in enumerate(centers):
            plt.scatter(center[0], center[1],
                        s=200, color='gray', marker='o')
            plt.scatter(center[0], center[1],
                        s=70, color='k', marker='$%d$' % i)

    if legend:
        plt.legend()

    plt.show()
```

```python
# 교재 코드가 아닌 ai로 생성
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
from sklearn.decomposition import PCA

def visualize_silhouette(cluster_list, X_features):
    fig, axes = plt.subplots(1, len(cluster_list),
                             figsize=(4 * len(cluster_list), 5))

    for i, n_clusters in enumerate(cluster_list):
        kmeans = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
        labels = kmeans.fit_predict(X_features)

        silhouette_avg = silhouette_score(X_features, labels)
        sample_silhouette_values = silhouette_samples(X_features, labels)

        ax = axes[i]
        y_lower = 10

        for j in range(n_clusters):
            cluster_values = sample_silhouette_values[labels == j]
            cluster_values.sort()

            size_cluster = len(cluster_values)
            y_upper = y_lower + size_cluster

            color = plt.cm.nipy_spectral(float(j) / n_clusters)

            ax.fill_betweenx(
                np.arange(y_lower, y_upper),
                0, cluster_values,
                facecolor=color, edgecolor=color, alpha=0.9
            )

            ax.text(-0.05, y_lower + 0.5 * size_cluster, str(j))
            y_lower = y_upper + 10

        ax.axvline(x=silhouette_avg, color='red', linestyle='--')
        ax.set_xlim([-0.1, 1])
        ax.set_yticks([])
        ax.set_xlabel('The silhouette coefficient values')
        ax.set_ylabel('Cluster label')
        ax.set_title(f'Number of Cluster: {n_clusters}\n'
                     f'Silhouette Score: {silhouette_avg:.3f}')

    plt.tight_layout()
    plt.show()


def visualize_kmeans_plot_multi(cluster_list, X_features):
    X_features = np.asarray(X_features)

    if X_features.shape[1] > 2:
        X_plot = PCA(n_components=2).fit_transform(X_features)
    else:
        X_plot = X_features

    fig, axes = plt.subplots(1, len(cluster_list),
                             figsize=(4 * len(cluster_list), 4))

    markers = ['o', 's', '^', 'x', '*', 'D', 'P']

    for i, n_clusters in enumerate(cluster_list):
        kmeans = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
        labels = kmeans.fit_predict(X_features)

        ax = axes[i]

        for j in range(n_clusters):
            mask = labels == j

            ax.scatter(
                X_plot[mask, 0],
                X_plot[mask, 1],
                marker=markers[j % len(markers)],
                edgecolor='k',
                label=f'Cluster {j}'
            )

        ax.set_title(f'Number of Cluster: {n_clusters}')
        ax.legend(fontsize=8)

    plt.tight_layout()
    plt.show()
```

```python
visualize_silhouette([2, 3, 4, 5], X_features_scaled)
visualize_kmeans_plot_multi([2, 3, 4, 5], X_features_scaled)
```
