### 1-1-2. 주어진 리스트에 있는 (x, y) 쌍에 대해 조건에 맞는 결과문 리스트를 출력해 주세요.(6점)

    *   5만원 미만 결제 시: 배송비 3000원

|   각 조건에 맞는 배송비를 `delivery_fee`에 정확하게 저장함 | 3점|
|   `total_price`와 `delivery_fee`를 이용하여 최종 결제 금액을 `final_price`에 정확하게 계산함 | 2점|

````python
### 이 셀은 수정/삭제하시면 안됩니다. 실행해주세요!

first_save = 10000   # 첫 달 저축액
increase = 5000      # 매달 늘어나는 금액
````

````python
###답안 작성###
month = 0
total = 0
save = first_save

while total < 200000:
    total += save
    save += increase
    month += 1

# 결과 출력 ( 아래 코드는 수정/삭제하지 말아주세요. )
print(f"목표 달성까지 {month}개월이 걸립니다.")
````

### 2-4-1. 이차원 리스트 계산.(3점)
아래 항목을 각각 계산하여 출력하세요.

1.   `sales`의 모든 원소를 계산이 가능한 형태로 바꾼 새 이차원 리스트 `sales_new` 생성
2.   각 회차별 총 판매량을 계산해 `round_total` 리스트에 저장
3.   각 구역별 최대 판매량을 계산해 `zone_max` 리스트에 저장

📌 조건

*   회차별 판매량은 가로 방향(한 행)에 나열되어 있음
*   구역별 판매량은 세로 방향(각 열)에 나열되어 있음
*   직접 값을 입력하면 0점 처리

💚 Test Case

*   출력

    회차별 총 판매량: [295, 390, 255]

    구역별 최대 판매량: [150, 130, 110]

✅ 평가 기준

모든 원소를 정수형으로 변환하여 `sales_new`를 생성한다.(1점)

행 단위로 합계를 구해 `round_total`을 생성한다.(1점)

열 단위로 최댓값을 구해 `zone_max`를 생성한다.(1점)

- 형변환 없이 정수 리스트를 새로 입력한 경우 `sales_new` 0점


````python
sales = [["120", "95", "80"],
         ["150", "130", "110"],
         ["100", "85", "70"]]

# sales_new 생성
sales_new = [[int(x) for x in row] for row in sales]

# 회차별 총 판매량 계산
round_total = [sum(row) for row in sales_new]

# 구역별 최대 판매량 계산
zone_max = [max(row[j] for row in sales_new) for j in range(len(sales_new[0]))]

# 결과 출력 ( 아래 코드는 수정/삭제하지 말아주세요. )
print(f"회차별 총 판매량: {round_total}")
print(f"구역별 최대 판매량: {zone_max}")
````

정수 배열 nums를 입력 받아,
배열에서 각 숫자보다 앞에 있는 숫자들 중, 자신보다 큰 숫자의 개수를 구하여 새로운 리스트에 저장하고 출력하세요.


📌 입력 형식

     배열 요소 (공백으로 구분된 정수)

     예)

     5 2 7 3

     

📌 출력 형식

    [0, 1, 0, 2]
     (설명:
     5 → 앞에 있는 숫자가 없으므로 0
     2 → 앞의 5가 더 크므로 1
     7 → 앞의 5 2중 더 큰 숫자가 없으므로 0
     3 → 앞의 5, 2, 7 중 5, 7이 더 크므로 2)

💚 Test Case

예시1)

Input: nums = [4, 1, 3, 2]

Output: [0, 1, 1, 2]

예시2)

Input: nums = [3,3]

Output: [0,0]

📌 조건

*   리스트 result 에 결과를 저장하고 최종 출력할 것
*   같은 인덱스 두 번 사용 금지
*   for 반복문 사용
*   리스트 인덱스 접근 사용
*   각 숫자보다 앞에 있는 원소만 탐색할 것
*   반드시 하나의 정답만 존재


✅ 평가 기준

| 항목                              | 점수 |
| ------------------------------- | -- |
| 이중 for 반복문 구조 구현                    | 3점 |
| 각 원소별 counting 로직 구현 | 2점 |
| 인덱스 중복 방지 (`j in range(i)` 등) | 1점 |
| 크기를 비교하는 조건문 구현             | 2점 |
| 리스트 형태로 반환                    | 2점 |

````python
nums = list(map(int, input().split()))
result = []

###답안 작성###
for i in range(len(nums)):
    count = 0
    for j in range(i):
        if nums[j] > nums[i]:
            count += 1
    result.append(count)
print(result)
````

### 3-3-1. 함수 구현.(7점)

정수 배열에서 1씩 증가하는 숫자가 연속해서 나타나는 구간 중 가장 긴 구간의 길이를 구하여 반환하세요.


📌 함수 형식 (반드시 유지)


```
def longestSequence(nums):
```

반환값: 가장 긴 연속 구간의 길이를 나타내는 정수



💚 Test Case

예시2) 연속 구간이 없는 경우

📌 조건
*   반드시 함수를 작성하고, 결과를 return하여 반환할 것
*   예시 1, 2의 리스트를 직접 대입하여 결과를 출력할 것
*   연속된 구간의 길이를 저장하는 변수를 사용할 것
*   리스트 인덱스 접근 사용
*   for 반복문을 사용
*   반드시 하나의 정답만 존재

✅ 평가 기준
| 항목                   | 점수 |
| -------------------- | -- |
| 적절한 변수 정의 및 초기화 여부     | 2점 |
| 이웃한 원소 사이 1씩 증가하는지 조건문 구현 | 2점 |
| 최댓값 갱신 로직 구현            | 2점 |
| 정수 형태를 반환하는 형식 및 예시 1,2 대입 | 1점 |

````python
###답안 작성###
def longestSequence(nums):
    count = 1
    max_count = 1

    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1] + 1:
            count += 1
        else:
            count = 1

        if count > max_count:
            max_count = count

    return max_count

nums1 = [-3, -2, -1, 4, 5, 10, 11]
nums2 = [-5, 0, 3, 7, 10]
print(longestSequence(nums1))
print(longestSequence(nums2))
````
