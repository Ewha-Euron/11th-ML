### 15. What is the result of the following expression?

(실행 후, 결과에 대한 주석 작성해주세요. ex. 출력 결과에 대한 이유)

```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
0 * np.nan #nan은 결측값이기 때문에 결과를 확정할 수 없어 nan이 된다. 
np.nan == np.nan #nan은 결측값, 즉 알 수 없는 값이기 때문에 어떤 값과 비교해도 같다고 인정하지 않아 False
np.inf > np.nan #nan은 알 수 없는 값이라서 비교하면 False 
np.nan - np.nan #nan이 포함된 계산. 결측값에서 결측값을 빼면 정확한 값을 알 수 없어서 False 
np.nan in set([np.nan]) #set 내에 같은 객체를 찾을 수 있어서 True
0.3 == 3 * 0.1 #부동소수점 오차 때문에 False
````
