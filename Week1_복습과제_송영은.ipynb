### 15. What is the result of the following expression?

(실행 후, 결과에 대한 주석 작성해주세요. ex. 출력 결과에 대한 이유)

```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
#  nan: nan이 포함된 산술 연산의 결과는 nan이다.
print(0 * np.nan)

# False: nan은 자기 자신과도 같다고 비교되지 않는다.
print(np.nan == np.nan)

# False: nan은 크기를 비교할 수 없는 값이다.
print(np.inf > np.nan)

#  nan: nan이 포함된 산술 연산의 결과는 nan이다.
print(np.nan - np.nan)

# True: set 안에 동일한 np.nan 객체가 존재한다.
print(np.nan in set([np.nan]))

# False: 부동소수점 표현의 오차 때문에 두 값이 정확히 같지 않다.
print(0.3 == 3 * 0.1)
````

### 21. Given a DataFrame of random numeric values:
```python
df = pd.DataFrame(np.random.random(size=(5, 3))) # this is a 5x3 DataFrame of float values
```

how do you subtract the row mean from each element in the row?

````python
# 랜덤 시드 고정
np.random.seed(2025)

df2 = pd.DataFrame(np.random.random(size=(5, 3)))
print('# before')
df2
````

````python
df2 = df2.sub(df2.mean(axis=1), axis=0)
print('# after')
df2
````
